## Leé un CSV y hacé una gráfica
Estos datos son inventados. Podés cambiarlos sin subir archivos personales.

```python
import pandas as pd
import matplotlib.pyplot as plt
from io import StringIO
datos = pd.read_csv(StringIO('mes,valor\nEnero,4\nFebrero,7\nMarzo,6'))
datos.plot.bar(x='mes', y='valor', legend=False)
plt.ylabel('Valor de ejemplo')
plt.show()
```
