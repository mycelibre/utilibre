## Read a CSV and draw a chart
This example uses invented data. You can edit it without uploading a personal file.

```python
import pandas as pd
import matplotlib.pyplot as plt
from io import StringIO
data = pd.read_csv(StringIO('month,value\nJan,4\nFeb,7\nMar,6'))
data.plot.bar(x='month', y='value', legend=False)
plt.ylabel('Example value')
plt.show()
```
